```python
import pandas as pd
import numpy as np

filepath = 'https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv'

df = pd.read_csv(filepath)

# Shape
print("----- SHAPE -----")
print(df.shape)
print()

# Column names
print("----- COLUMN NAMES -----")
print(df.columns.tolist())
print()

# Datatypes
print("----- DATA TYPES -----")
print(df.dtypes)
print()

# Missing values
print("----- MISSING VALUES -----")
print(df.isnull().sum())
print()

# Head
print("----- FIRST 5 ROWS -----")
print(df.head(5))
print()

# Check and remove duplicates
print("----- DUPLICATES -----")
print("Before removing:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("After removing:", df.duplicated().sum())
print()

# Rename column
df.rename(columns={'Title': 'Movie_Title'}, inplace=True)

print("----- RENAMED COLUMNS -----")
print(df.columns.tolist())
print()

# Get unique genres
print("----- UNIQUE GENRES -----")
print(df.Genre.unique())
print()

# Number of genres per movie
df['Genre_Count'] = df['Genre'].str.split(',').apply(len)

print("----- NUMBER OF GENRES PER MOVIE -----")
print(df[['Movie_Title', 'Genre', 'Genre_Count']])
print("\n")
# Handle missing values in 'Revenue (Millions)' and 'Metascore'

df['Revenue (Millions)'] = df['Revenue (Millions)'].fillna( df['Revenue (Millions)'].mean())

df['Metascore'] = df['Metascore'].fillna(df['Metascore'].mean())

print("----- MISSING VALUES AFTER HANDLING -----")
print(df.isnull().sum())
print("\n")

# Top 5 movies by Rating
print("----- TOP 5 MOVIES BY RATING -----")
print(df[['Movie_Title', 'Rating']].sort_values(by='Rating', ascending=False).head(5))
print("\n")

# Movies with highest/lowest revenue
print("----- MOVIES WITH HIGHEST REVENUE -----")
print(df[['Movie_Title', 'Revenue (Millions)']].sort_values(by='Revenue (Millions)', ascending=False).head(5))
print("\n")

print("----- MOVIES WITH LOWEST REVENUE -----")
print(df[['Movie_Title', 'Revenue (Millions)']].sort_values(by='Revenue (Millions)', ascending=True).head(5))
print("\n")

# Avg. rating by Director
print("----- AVERAGE RATING BY DIRECTOR -----")
print(df.groupby('Director')['Rating'].mean().sort_values(ascending=False))
print("\n")

# Movies per year
print("----- MOVIES PER YEAR -----")
print(df['Year'].value_counts())
print("\n")

# Movies with rating > 8 & revenue > 100M
print("----- MOVIES WITH RATING > 8 & REVENUE > 100M -----")
print(df[(df['Rating'] > 8) & (df['Revenue (Millions)'] > 100)])
print("\n")

# Group by Director: avg Rating, Revenue
print("----- GROUP BY DIRECTOR: AVG RATING, REVENUE -----")
print(df.groupby('Director')[['Rating', 'Revenue (Millions)']].mean())
print("\n")

# Group by Year: count, avg Rating
print("----- GROUP BY YEAR: COUNT, AVG RATING -----")
print(df.groupby('Year')['Rating'].agg(['count', 'mean']))
print("\n")

# Create Profitability column
print("----- CREATED PROFITABILITY COLUMN -----")
df['Profitability'] = df['Revenue (Millions)'] / df['Runtime (Minutes)']
print(df[['Movie_Title', 'Revenue (Millions)', 'Runtime (Minutes)', 'Profitability']])
print("\n")

# Top 10 most profitable movies

print("----- TOP 10 MOST PROFITABLE MOVIES -----")
print(df[['Movie_Title', 'Revenue (Millions)', 'Runtime (Minutes)', 'Profitability']].sort_values(by='Profitability', ascending=False).head(10))
print("\n")

# Create a column that categorizes movies by rating:, a) Rating < 6: Poor, b) 6 ≤ Rating < 7.5: Average, c) Rating ≥ 7.5: Excellent

print("----- RATING CATEGORY -----")

df['Rating_Category'] = pd.cut(df['Rating'],bins=[0, 6, 7.5, float('inf')],labels=['Poor', 'Average', 'Excellent'],right=False)

print(df[['Movie_Title', 'Rating', 'Rating_Category']])
print("\n")

# Custom function to calculate movie score

def movie_score(row):return row['Rating'] * 10 + row['Revenue (Millions)']

# Apply function to dataset
df['Score'] = df.apply(movie_score, axis=1)

print("----- MOVIE SCORES -----")
print(df[['Movie_Title', 'Rating', 'Revenue (Millions)', 'Score']])

```
